## Primer Ejercicio
Serie Fibonacci

```python
numero_actual = 0
numero_anterior = 1
while numero_actual < 20:
    numero = numero_actual + numero_anterior
    numero_anterior = numero_actual
    numero_actual = numero
    print(numero)
```

## TAREA
Generar e imprimir el tríangulo de Pascal

```python
def print_pascal_row(row):
    print(*row, sep = " ")


def generar_triangulo_pascal(limit):
    pascal_row = [1]
    print_pascal_row(pascal_row)
    pascal_row = [1, 1]
    print_pascal_row(pascal_row)
    next_pascal_row = []
    for i in range(1, limit):
        next_pascal_row.append(1)
        for index, val in enumerate(pascal_row):
            if index < len(pascal_row)-1:
                new_index = index+1
                number_value = val + pascal_row[new_index]
                next_pascal_row.append(number_value)
        next_pascal_row.append(1)        
        print_pascal_row(next_pascal_row)
        pascal_row = next_pascal_row
        next_pascal_row = []


limit = 6
generar_triangulo_pascal(limit)
```
