# Tarea Diagrama Herzsprung-Russell

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Luminosidad en unidades de Luminosidad solar


```python
id, M, L, T= np.loadtxt('datos_camilo/tabla_camilo_HR_LS.txt', unpack=True)
```

```python
ax = plt.axes()
ax.set_facecolor("black")
plt.rcParams.update({'font.size': 14})
plt.plot(T,L,"y*")
# plt.grid()
plt.xscale('log')
plt.yscale('log')
plt.xlabel("Temperatura (Kelvin)")
plt.ylabel("Luminosidad (L/Lsun)")
plt.title("Diagrama Herzsprung-Russell")
plt.gca().invert_xaxis()
plt.show()
```

### Luminosidad en unidades de Vatios (W)



```python
id, M, L, T= np.loadtxt('datos_camilo/tabla_camilo_HR_W.txt', unpack=True)
```

```python
ax = plt.axes()
ax.set_facecolor("black")
plt.rcParams.update({'font.size': 14})
plt.plot(T,L, "*", color='orange')
# plt.grid()
plt.xscale('log')
plt.yscale('log')
plt.xlabel("Temperatura (Kelvin)")
plt.ylabel("Luminosidad (Vatios)")
plt.title("Diagrama Herzsprung-Russell")
plt.gca().invert_xaxis()
plt.show()
```
